**Regression Problem**

Problem statement : House Price Prediction


*   Predict the house price based on : House size(sqft), number of rooms




```python
import torch
import torch.nn as nn
import torch.optim as optim
# optim provides optimizers that update model
# weights so your neural network learns from data.

# Creating data sets
X = torch.tensor([
    [800,2],
    [1000,3],
    [1200,3],
    [1500,4],
    [1800,4]
], dtype=torch.float32)

y = torch.tensor([[200],[260],[300],[360],[420]], dtype=torch.float32)


# Normalizing inputs
# Statistical way of normalization, range will be [0, 1]
# Please note that, we haven't used the normalization techniques available in NN Yet
X = X / X.max(dim=0).values
y = y / y.max()



# Define model :
"""
- Sequential is a container offered by nn submodule, to define fully connected layers activation functions & forward propagation steps, all these steps execute step by step
- nn.Linear(2,8) means, passing 2 inputs(size, no. of bedroom) to a hidden layer with 8 neurons
- nn.ReLU() ==> Using ReLU activation function
- nn.Linear(8,1) ==> Output layer with 1 neuron

Also note :
- for a regression problem action function for o/p layer is Linear activation function. Hence not defined any activation function here for output layer.

"""
model = nn.Sequential(
    nn.Linear(2,8),
    nn.ReLU(),
    nn.Linear(8,1)
)


# Loss & Optimizer
"""
For a regression problem, we have to get the Loss.
"""
criterion = nn.MSELoss() # Regression. Hence MSE. Could use MAE or Huber Loss.
optimizer = optim.Adam(model.parameters(), lr=0.01) # Using Adam Optimizer, it will optimizer both speed and Learning rate.


# Traning Loop
for epoch in range(500): # 0 - 449 times ML model will iterate
    pred = model(X) # passing i/p data 'X' to our model, we will get a predicted value, passig it to 'pred' (pred is nothing but Y^)
    loss = criterion(pred,y) # Using MSELoss() to calculate Loss

    optimizer.zero_grad() # Setting gradient value to '0' before starting every iteration (intentional, otherwise GD value will accumulate in each iteration)
    loss.backward() # passing loss to backward propagation, generally BP starts from Loss
    optimizer.step() # Adjust Weights & Biases (W_new = W_old - LR*(dL/dW_old) & B_new = WBold - LR*(dL/dB_old))

    # Instead of printing all epochs, print every 50th epoc
    if epoch % 50 == 0:
        print(f"Epoch {epoch}, Loss {loss.item():.4f}")



# Test Prediction
test = torch.tensor([[1600,4]], dtype=torch.float32)
test = test / X.max(dim=0).values

prob = model(test)
print("Pass probability:", prob.item())
print("Prediction:", 1 if prob>0.5 else 0)

print("Predicted price, Y :", model(test)*420) # Initially, we divided Y value with Y.max. Hence now multiplying with same no. to get actual price.
```
