```python
"""
==> NN offers different types of layers as below:
1. Linear (for Artificial Neural Network)
2. Conv2D (for image processing)
3. LSTM (for sequence - sequence data)
4. RNN (for sequence - sequence data)

We are dealing with Artificial Neural Network(ANN) in this program. Hence using Linear layers.

==> NN will offer
      - activations
      - loss functions
      - backward propagation (train(), eval())
      - Optimizers (torch.optim)

# fc means fully conencted layer
# self.fc1, self.fc2 means these fully connected layers wrapper with current instance of class

Linear() represents a fully connected layer in NN
self.fc1 = nn.Linear(3, 5) ; 3 nodes in input layer & 5 in hidden layer
self.fc2 = nn.Linear(5, 2) ; 5 nodes from hidden layer & 2 nodes in output layer
==> We should mention similar number of nodes in hidden layer in both the places

"""

# importing modules torch, nn(neural network): nn is a sub module in main module torch
import torch
import torch.nn as nn

class CalculateParams(nn.Module):
  def __init__(self): # Constructor used to define layers and parameters
    super().__init__() # Calls parent nn.Module constructor, it will register our model so PyTorch can track prameters, load weights etc.
    self.fc1 = nn.Linear(3, 5) # Fully connected layer1
    self.fc2 = nn.Linear(5, 2) # Fully connected layer2

  # forward() defines how a input flows through a network
  # we never call it directly, it will be automatically called when we do 'output = Model(input)'
  def forward(self, x):
    x = self.fc1(x)
    x = self.fc2(x)

    return x

model = CalculateParams()

print()
total_params = sum(p.numel() for p in model.parameters()) # p.numel() returns number of elements in that tensor
print("Total parameters:", total_params)


"""
for name, param in model.named_parameters():
    print("Name:", name)
    print("Shape:", param.shape)
    print("Values:\n", param)
    print("-" * 40)



Name: fc1.weight
Shape: torch.Size([5, 3])
Values:
 Parameter containing:
tensor([[-0.0385, -0.2991, -0.5688],
        [-0.3523,  0.4376, -0.0818],
        [-0.0935, -0.1909,  0.1437],
        [-0.3362, -0.5095,  0.0213],
        [-0.3732,  0.0166, -0.1583]], requires_grad=True)
----------------------------------------
Name: fc1.bias
Shape: torch.Size([5])
Values:
 Parameter containing:
tensor([-0.3382, -0.4032, -0.0229, -0.4300, -0.2626], requires_grad=True)
----------------------------------------
Name: fc2.weight
Shape: torch.Size([2, 5])
Values:
 Parameter containing:
tensor([[ 0.1248, -0.3440,  0.2765, -0.3777,  0.1273],
        [ 0.0524,  0.3726,  0.0515,  0.2485,  0.0469]], requires_grad=True)
----------------------------------------
Name: fc2.bias
Shape: torch.Size([2])
Values:
 Parameter containing:
tensor([0.0380, 0.3034], requires_grad=True)
"""
```
