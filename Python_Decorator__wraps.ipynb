```python
from functools import wraps

"""
Points to remember :

Step 1 :
======
First we need to understand what is the meta data of function say_hello()

| Attribute         | Meaning               |
| ----------------- | --------------------- |
| `__name__`        | Function name         | say_hello (as per below code)
| `__doc__`         | Docstring             | Greet the user (as per above code)
| `__module__`      | Module name           |
| `__annotations__` | Type hints            |
| `__wrapped__`     | Reference to original |

Step 2 :
======
Interpretor flow will start from top to bottom

when it comes to def mydec(func)
✔ It creates a function object
✔ Stores it in memory
✔ Assigns name my_dec to it
❌ Does NOT run its body

Step 3 :
======
Now, at
@my_dec
def say_hello():

Memory allocated (and also decorated immediately)
✔ Docstring stored
✔ Body NOT executed

Step 4 :
======
Now, Applies decorator (IMPORTANT)
So:

✔ my_dec() EXECUTES
✔ wrapper is created
✔ returned
✔ assigned to say_hello

Now say_hello → wrapper


Step 5 :
======
Print statements
    - Understand we are not calling say_hello method as say_hello()
    - We are just referencing the function object and its meta data
    - Hence below 2 lines of code never execute inside def wrapper(*args, **kwargs):
        print("Something")
        return func(*args, **kwargs)
    - Try using "pass" insted of above 2 lines then you will understand, you will get same output

print(say_hello.__name__)
print(say_hello.__doc__)


Finally :
=======

Metadata becomes part of wrapper object
✔ It is NOT temporary
✔ It does NOT expire (until Function reassigned, function deleted, program exits)
✔ It is NOT recalculated

"""

def my_dec(func):
  @wraps(func)
  def wrapper(*args, **kwargs):
    print("Something")
    return func(*args, **kwargs)

  return wrapper

@my_dec
def say_hello(): # python internally does "say_hello = my_dec(say_hello)"
  "Greet the user"
  print("Hello !")


print(say_hello.__name__) # this is the name of the function reference in meta data
print(say_hello.__doc__) # this is the docstring inside function reference

# print statement inside wrapper & say_hello() will execute when you call below function
#say_hello()
```
