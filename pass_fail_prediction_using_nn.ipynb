**Classification Problem (Pass/Fail Prediction)**

🎯 Problem

Predict if a student Passes (1) or Fails (0) based on:

Hours studied

Attendance %

```python
import torch
import torch.nn as nn
import torch.optim as optim


# Dataset
X = torch.tensor([
    [1,40],
    [2,50],
    [3,60],
    [4,65],
    [5,75],
    [6,85]
], dtype=torch.float32)

y = torch.tensor([[0],[0],[0],[1],[1],[1]], dtype=torch.float32)

X = X / X.max(dim=0).values


# Model
model = nn.Sequential(
    nn.Linear(2,8),
    nn.ReLU(),
    nn.Linear(8,1),
    nn.Sigmoid()
)


# Loss & Optimizer
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)


# Training
for epoch in range(500):
    pred = model(X)
    loss = criterion(pred,y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch % 50 == 0:
        print(f"Epoch {epoch}, Loss {loss.item():.4f}")


# Test Prediction
test = torch.tensor([[4,70]], dtype=torch.float32)
test = test / X.max(dim=0).values

prob = model(test)
print("Pass probability:", prob.item())
print("Prediction:", 1 if prob>0.5 else 0)
```
